### Types of Function in Seaborn
- Figure Level
- Axis Level

### Main Classification
**1. Relational Plots** 

**2. Distribution Plots**

**3. Categorical Plots**

**4. Regresson Plots**

**5. Matrix Plots**

**6. Multiplots**

```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

### 1. Relational Plots
- to see the statisical relation between 2 or more variables
- Brvariate Analysis

**Plots under This section**
- scatterplot
- lineplot

**figure level fuction -  `relplot()`**

```python
tips = sns.load_dataset('tips')
tips.head()
```

```python
# scatter plot -> axes level function
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='sex')
```

```python
# figure level fuction
sns.relplot(data=tips, x='total_bill', y='tip', kind='scatter', hue='sex', style='time', size='size')
```

```python
 # line plot
stocks = pd.read_csv('data sets/stocks.csv')
stocks.head()
```

```python
# AXES LEVEL function
sns.lineplot(data=stocks, x='Date', y='MSFT')
```

```python
sns.relplot(data=stocks, x='Date', y='MSFT', kind='line')
```

```python
# facet plot - fig level, work with relplot
sns.relplot(data=tips, x='total_bill', y='tip', kind='scatter', col='sex')
```

```python
sns.relplot(data=tips, x='total_bill', y='tip', kind='scatter', col='sex', row='smoker')
```

- Are there outliers ?

**Plots under distribution plot**
- hisplot
- kdeplot
- rugplot

**figure level fuction -  `displot()`**

```python
sns.displot(data=tips, x='tip', kind='hist', hue='sex', element='step')
```

```python
sns.displot(data=tips, x='tip', kind='hist', col='sex')
```

```python
# kdeplot 
sns.kdeplot(data=tips, x='total_bill', hue='sex')
```

### 3. Matric Plot 
- Heatmap
- Clustermap

```python
# heatmap 
deliveries = pd.read_csv('data sets/IPL_Ball_by_Ball_2008_2022.csv')
deliveries.head()
```

```python
# filter sixes rows only
sixes_df = deliveries[(deliveries['ballnumber'].isin([1,2,3,4,5,6])) & (deliveries['batsman_run'] == 6)]
sixes_df.head()
```

```python
# Create a sixes grid using pivot table 
sixes_grid = sixes_df.pivot_table(
    index='overs',
    columns='ballnumber',
    values='batsman_run',
    aggfunc='count'
)
sixes_grid.head()
```

```python
plt.figure(figsize=(10,15))
sns.heatmap(sixes_grid, cmap='winter', annot=True, linewidths=0.5)
plt.yticks(deliveries['overs'].unique(), list(range(1,21)))
plt.xticks(np.arange(0,6),list(range(1,7)))
plt.xlabel('Ball Number')
plt.ylabel('Over')
plt.title('Sixes')
```

```python
# Clustermap 
iris = pd.read_csv('data sets/iris.csv')
iris
```

```python
sns.clustermap(iris.iloc[:, [0,1,2,3]])
```

### 4. Categorical Plots
**Categorical Scatter Plot**
- Stripplot
- Swarmplot

**Categorical Distribution Plot**
- Boxplot
- violinplot
  
**Categorical Estimate Plot -> for central tendency**
- Barplot
- Pointplot
- Countplot

**Figure level function -> `catplot()`**

```python
# stripplot
sns.stripplot(data=tips, x='day', y='total_bill')
```

```python
sns.stripplot(data=tips, x='day', y='total_bill', jitter=False)
```

```python
sns.catplot(kind='strip', data=tips, x='day', y='total_bill', jitter=0.1, hue='sex')
```

```python
# swarmplot 
sns.catplot(kind='swarm',data=tips, x='day', y='total_bill', hue='sex')
```

```python
sns.swarmplot(data=tips, x='day', y='total_bill')
```

```python
# Box Plot 
sns.boxplot(data=tips, y='total_bill')
```

```python
sns.boxplot(data=tips, x='sex', y='total_bill')
```

```python
sns.catplot(data=tips, x='day', y='total_bill', kind='box', hue='sex')
```

```python
# violinplot 
sns.violinplot(data=tips, x='day', y='total_bill')
```

```python
sns.catplot(data=tips, x='day', y='total_bill', kind='violin', hue='sex')
```

```python
sns.catplot(data=tips, x='day', y='total_bill', kind='violin', hue='sex', split=True)
```

```python
# barplot
sns.barplot(data=tips, x='sex', y='total_bill')
```

```python
sns.barplot(data=tips, x='sex', y='total_bill', errorbar=None)
```

```python
sns.catplot(data=tips, x='sex', y='total_bill', errorbar=None, kind='bar', hue='smoker')
```

```python
# point plot 
sns.catplot(data=tips, x='sex', y='total_bill', errorbar=None, kind='point', hue='smoker')
```

```python
# countplot
sns.countplot(data=tips, x='sex')
```

```python
sns.catplot(data=tips, x='sex', hue='day', kind='count')
```

```python
# faceting plot 
sns.catplot(data=tips, x='sex', y='total_bill', kind='box', hue='smoker', col='day', col_wrap=2)
```

 ### 6. Multiplots

```python
# FacetGrid 
g = sns.FacetGrid(data=tips,col='day', row='time')
g.map(sns.violinplot,'sex','total_bill')
```

#### Pairplot (Beta Of Pair Grid)

```python
iris = iris.drop('Id',axis=1)
```

```python
iris.head()
```

```python
sns.pairplot(data=iris)
```

```python
sns.pairplot(data=iris, hue='Species')
```

#### PairGrid (Baap of Pair Plot)

```python
g = sns.PairGrid(iris,hue='Species')
g.map(sns.scatterplot)
```

```python
g = sns.PairGrid(iris,hue='Species')
g.map_diag(sns.histplot, element='step')
g.map_offdiag(sns.scatterplot)
```

```python
g = sns.PairGrid(iris,hue='Species')
g.map_diag(sns.histplot, element='step')
g.map_upper(sns.scatterplot)
g.map_lower(sns.kdeplot)
```

```python
# vars
g = sns.PairGrid(iris,hue='Species', vars=['SepalLengthCm', 'PetalWidthCm'], height=5, aspect=2)
g.map_diag(sns.histplot, element='step')
g.map_upper(sns.scatterplot)
g.map_lower(sns.kdeplot)
```

#### Jointplot (beta of Joint Grid)

```python
sns.jointplot(data=tips, x='total_bill', y='tip', hue='sex')
```

```python
sns.jointplot(data=tips, x='total_bill', y='tip', kind='kde', hue='sex')
```

#### JointGrid (papa of Joint plot)

```python
g = sns.JointGrid(data=tips, x='total_bill', y='tip', hue='sex')
g.plot(sns.scatterplot, sns.kdeplot)
```
