### Types of Data
- Numerical Data
- Categorical Data

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### 1. 2D Line Plot
- Bivarate Analysis
- Categorical vs Numerical and Numerical vs Numerical
- Use Case - Time Series Data

```python
batsman = pd.read_csv('Data Sets/sharma-kohli.csv')
batsman
```

```python
# ploting sigle plot
plt.plot(batsman['index'], batsman['RG Sharma'])
```

```python
# ploting multiple plots
plt.plot(batsman['index'], batsman['RG Sharma'])
plt.plot(batsman['index'], batsman['V Kohli'])
```

```python
# lables and title
plt.plot(batsman['index'], batsman['RG Sharma'])
plt.plot(batsman['index'], batsman['V Kohli'])

plt.title('Rohit Sharma Vs Virat Kohli Career Comparsion')
plt.xlabel('Seasons')
plt.ylabel('Runs')
plt.show()
```

```python
# colors(hex) and line(width and style) and marker(size)
plt.plot(batsman['index'], batsman['RG Sharma'], color='red', linestyle='dotted', marker='o', markersize='4')
plt.plot(batsman['index'], batsman['V Kohli'], color='black', linestyle='dashed', marker='D')

plt.title('Rohit Sharma Vs Virat Kohli Career Comparsion')
plt.xlabel('Seasons')
plt.ylabel('Runs')
plt.show()
```

```python
# legends
plt.plot(batsman['index'], batsman['RG Sharma'], marker='o', markersize='4', label='Rohit Sharma')
plt.plot(batsman['index'], batsman['V Kohli'], marker='o', markersize='4', label='Virat Kohli')

plt.title('Rohit Sharma Vs Virat Kohli Career Comparsion')
plt.xlabel('Seasons')
plt.ylabel('Runs')

plt.legend()

plt.show()
```

```python
# grid
plt.plot(batsman['index'], batsman['RG Sharma'], marker='o', markersize='4', label='Rohit Sharma')
plt.plot(batsman['index'], batsman['V Kohli'], marker='o', markersize='4', label='Virat Kohli')

plt.title('Rohit Sharma Vs Virat Kohli Career Comparsion')
plt.xlabel('Seasons')
plt.ylabel('Runs')

plt.legend()
plt.grid()

plt.show()
```

### 2. Scatter Plot
- Bivarate Analysis
- Numerical vs Numerical
- Use Case - Finding correlation

```python
better = pd.read_csv('Data Sets/batter.csv')
better = better.head(50)
better.head()
```

```python
# basic scatter plot 
plt.scatter(better['avg'], better['strike_rate'])

plt.title('Avg and SR analysis of Top 50 IPL Batsman')
plt.xlabel('Average')
plt.ylabel('SR')
```

```python
tips = sns.load_dataset('tips')
tips
```

```python
# size  
plt.scatter(tips['total_bill'], tips['tip'], s=tips['size']*20)
```

```python
# scatter plot using plt.plot()
plt.plot(tips['total_bill'], tips['tip'], 'o') # its faster than plt.scatter()
```

### 3. Bar Chart 
- Bivarate Analysis
- Numerical vs Categorical
- Use Case - Aggregate Analysis of groups

```python
# simple bar chart
colors = ['Red', 'Blue', 'Green', 'Pink', 'Yellow']
children = [9, 7, 10, 15, 8]

plt.bar(colors, children)

plt.xlabel('Colors')
plt.ylabel('Children')
```

```python
# horizontal bar chart 
plt.barh(colors, children)

plt.xlabel('Children')
plt.ylabel('Colors')
```

```python
batsman_runs = pd.read_csv('Data Sets/batsman_season_record.csv')
batsman_runs
```

```python
# multiple bar charts
plt.bar(np.arange(batsman_runs.shape[0]) - 0.2, batsman_runs['2015'], width=0.2, label = '2015') 
plt.bar(np.arange(batsman_runs.shape[0]), batsman_runs['2016'], width=0.2,label = '2016') 
plt.bar(np.arange(batsman_runs.shape[0]) + 0.2, batsman_runs['2017'], width=0.2, label = '2017') 

# use to replace xlables values
plt.xticks(np.arange(batsman_runs.shape[0]), batsman_runs['batsman'])

plt.title('Top 5 Batsman Runs Scored in three different IPL Seasons')
plt.xlabel('Batsman')
plt.ylabel('Runs')

plt.legend()

plt.show()
```

```python
# a problem
colors = ['Red RedRedRed', 'Blue', 'Green Green Green', 'Pink Pink Pink Pink', 'Yellow']
children = [9, 7, 10, 15, 8]

plt.bar(colors, children)
```

```python
# soulation
colors = ['Red RedRedRed', 'Blue', 'Green Green Green', 'Pink Pink Pink Pink', 'Yellow']
children = [9, 7, 10, 15, 8]

plt.bar(colors, children)

plt.xticks(rotation = 80)
plt.show()
```

```python
# stacked bar chart
plt.bar(batsman_runs['batsman'], batsman_runs['2015'], label = '2015')
plt.bar(batsman_runs['batsman'], batsman_runs['2016'], bottom=batsman_runs['2015'], label = '2016') 
plt.bar(batsman_runs['batsman'], batsman_runs['2017'], bottom=(batsman_runs['2015'] + batsman_runs['2016']), label = '2017') 

plt.legend()
```

### 5. Pie Chart
- Bivarate/ Univarate Analysis
- Categorical vs Numerical
- Use Case - To find contribution on a standred scale

```python
df = pd.read_csv('Data Sets/gayle-175.csv')
df
```

```python
df = df[df['batsman_runs'] != 0]
df
```

```python
plt.pie(df['batsman_runs'], labels=df['batsman'], autopct = '%0.1f%%')
plt.show()
```

```python
# explode shadow
plt.pie(df['batsman_runs'], labels=df['batsman'], autopct = '%0.1f%%', explode=[0.1,0,0,0,0], shadow=True)
plt.show()
```
