```python
# write a function that can return the track record of 2 teams against each other
def track_recoard(team1, team2): 
    
    condition = ((ipl['Team1'] == team1) & (ipl['Team2'] == team2)) |  ((ipl['Team1'] == team2) & (ipl['Team2'] == team1))
    
    head_to_head = ipl[condition]
    
    total_matches = len(head_to_head)
    team1_wins = (head_to_head['WinningTeam'] == team1).sum()
    team2_wins = (head_to_head['WinningTeam'] == team2).sum()
    
    no_result_or_draw = total_matches - (team1_wins + team2_wins)
    
    return {
        'Total Matches': total_matches,
        f'{team1} Wins': team1_wins,
        f'{team2} Wins': team2_wins,
        'No Result/Draw': no_result_or_draw
    }
```

```python
print(track_recoard("Rajasthan Royals", "Royal Challengers Bangalore"))
```

```python
# find ranking of each movie in the group according to IMDB score
def rank_movie(group): 
    group['genre_rank'] = group['IMDB_Rating'].rank(ascending=False)
    return group
```

```python
# find normalized IMDB rating group wise
def normal(group): 
    group['Norm_Rating'] = (group['IMDB_Rating'] - group['IMDB_Rating'].min())/(group['IMDB_Rating'].max() - group['IMDB_Rating'].min())
    return group
```

### Joins

```python
# inner join (Full join)
students.merge(regs, how='inner', on='student_id')
```

```python
# left joins 
courses.merge(regs, how='left', on='course_id')
```

```python
# right join 
new_st = pd.DataFrame({
    'student_id' : [26,27,28], 
    'name' : ['Dalchand', 'Nandini', 'Akhil'], 
    'partner' : [27, 26, 29]
})
students = pd.concat([students, new_st], ignore_index=True)
```

```python
students.merge(regs, how='right', on='student_id')
```

```python
# outer join 
students.merge(regs, how='outer', on='student_id').tail(10)
```

```python
# 1. find total revenue generated
regs.merge(courses, how='inner', on='course_id')['price'].sum()
```

```python
# 2. find month by month revenue
pd.concat([reg_month1, reg_month2], keys=['Nov', 'Dec']).reset_index().merge(courses, how='inner', on='course_id').groupby('level_0')['price'].sum()
```

```python
# 3. Print the registration table
# cols -> name -> course -> price
regs.merge(students, on='student_id').merge(courses, on='course_id')[['name', 'course_name', 'price']]
```

```python
# 4. Plot bar chart for revenue/course
regs.merge(courses, on='course_id').groupby('course_name')['price'].sum().plot(kind='bar')
```

```python
# 5. find students who enrolled in both the months
both_months_St_idx = np.intersect1d(reg_month1['student_id'], reg_month2['student_id'])

students[students['student_id'].isin(both_months_St_idx)]
```

```python
# 6. find course that got no enrollment
course_list_id  = np.setdiff1d(courses['course_id'], regs['course_id'])
courses[courses['course_id'].isin(course_list_id)]
```

```python
# 7. find students who did not enroll into any courses
st_list_id  = np.setdiff1d(students['student_id'], regs['student_id'])
students[students['student_id'].isin(st_list_id)]
```

```python
(students[students['student_id'].isin(st_list_id)].shape[0] / len(students) ) * 100 # in percent
```

```python
# 8. Print student name -> partner name for all enrolled students
# self join 
students.merge(students, how='inner', left_on='partner', right_on='student_id')[['name_x', 'name_y']]
```

```python
# 9. find top 3 students who did most number enrollments
regs.merge(students, on='student_id').groupby(['student_id', 'name'])['name'].count().sort_values(ascending=False).head(3)
```

```python
# 10. find top 3 students who spent most amount of money on courses
regs.merge(students, on='student_id').merge(courses, on='course_id').groupby(['student_id', 'name'])['price'].sum().sort_values(ascending=False).head(3)
```

### **pivot**  <h4>Long data to wide format</h4>

```python
tips = sns.load_dataset('tips')
tips.sample(5)
```

```python
# find avg bill of smoker and no smoker male , female bill 

# way-1
tips.groupby(['sex', 'smoker'])[['total_bill']].mean().unstack()
```

```python
# way -2 pivot 
tips.pivot_table(index='sex', columns='smoker', values='total_bill', observed=True)
```

```python
# aggfunc
tips.pivot_table(index='sex', columns='smoker', values='total_bill', observed=True, aggfunc='sum')
```

```python
# all numeric cols 
tips.select_dtypes(include='number').pivot_table(index=tips['sex'], columns=tips['smoker'], observed=True)
```

```python
# multidimension
tips.pivot_table(index=['sex', 'smoker'], columns=['day', 'time'], values=['total_bill', 'tip'], observed=True)
```

```python
# margins 
tips.pivot_table(index='sex', columns='smoker', values='total_bill', observed=True, aggfunc='sum', margins=True)
```

## **Vectorized String Operation**

```python
titanic = pd.read_csv('Data sets/titanic.csv')
titanic.head()
```

```python
# common functions 
# lower/upper/capitalize/title 
titanic['Name'].str.capitalize()
```

```python
# len -> fint the long name person name 
titanic['Name'].str.len().max()
```

```python
titanic['Name'][titanic['Name'].str.len() == 82].values[0]
```

```python
# strip 
"          dalchand yogi     "
```

```python
"          dalchand yogi     ".strip()
```

```python
# split
titanic['SurName'] = titanic['Name'].str.split(',').str.get(0)
titanic.head()
```

```python
titanic[['Title', 'FirstName']] =  titanic['Name'].str.split(',').str.get(1).str.strip().str.split(' ', n=1, expand=True)
titanic.head()
```

```python
# count the Mr. and other 
titanic['Title'].value_counts()
```

```python
# replace 
titanic['Title'] = titanic["Title"].str.replace('Ms.', 'Miss.')
titanic['Title'] = titanic["Title"].str.replace('Mlle.', 'Miss.')
```

```python
titanic['Title'].value_counts()
```

```python
# filtering 
# startswith, endswith, isdigit, isalpha
titanic[titanic["FirstName"].str.endswith('A')]
```

```python
# applying regex 
# find the SurName that start ans end with vowel 
titanic[titanic['SurName'].str.contains('^[aeiouAEIOU].+[aeiouAEIOU]$')]
```

### DatetimeIndex Object 
A collection of pandas timestamps.

```python
# from sring 
pd.DatetimeIndex(['2026/9/9', '2026/9/10', '2026/9/11'])
```

```python
# using datetime 
pd.DatetimeIndex([dt.datetime(2026,9,9), dt.datetime(2026,9,10),dt.datetime(2026,9,11)])
```

```python
# using pd.timestamp 
dtx = pd.DatetimeIndex([pd.Timestamp(2026,9,9), pd.Timestamp(2026,9,10),pd.Timestamp(2026,9,11)])
dtx
```

```python
# use datetime index as series 
pd.Series([1,2,3], index=dtx)
```

### date_range() function

```python
# generate daily dates in range 
pd.date_range(start='2026/9/1', end='2026/9/10')
```

```python
pd.date_range(start='2026/9/1', end='2026/9/10', freq='2D') # frequency means gaf of days
```

```python
# B --> business days (Monday to Friday)
pd.date_range(start='2026/9/1', end='2026/9/10', freq='B')
```

```python
# H --> Hourly 
pd.date_range(start='2026/9/1', end='2026/9/2', freq='h')
```

```python
# using periods (number of results)
pd.date_range(start='2026/9/1', periods=20)
```
