##### **Q96. Find the top 3 customers within each region based on sales.**

```python
sales.groupby(['Region', 'Customer Name'])['Sales'].sum().groupby(level='Region').nlargest(3).droplevel(0)
```

##### **Q98. Find the worst-performing sub-category within each category based on profit.**

##### **Q99. Calculate each category's percentage contribution to total sales.**

```python
category_sales = sales.groupby('Category')['Sales'].sum()
category_percentage = (category_sales / sales['Sales'].sum()) * 100

# Format as a clean DataFrame
category_summary = pd.DataFrame({
    'Total Sales': category_sales,
    'Percentage (%)': category_percentage.round(2)
}).sort_values('Total Sales', ascending=False)

category_summary
```
