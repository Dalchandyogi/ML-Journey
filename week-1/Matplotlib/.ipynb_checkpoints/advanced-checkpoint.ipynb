### Annotations 

```python
betters = pd.read_csv('Data Sets/batter.csv')
betters.head()
```

```python
sample_df = betters.head(100).sample(25, random_state=5)
sample_df
```

```python
plt.figure(figsize=(15,8))

plt.scatter(sample_df['avg'], sample_df['strike_rate'], s=sample_df['runs'])



# for giving text to points
for i in range(sample_df.shape[0]):
    plt.text(sample_df['avg'].values[i], sample_df['strike_rate'].values[i], sample_df['batter'].values[i])

# for horizontal line 
plt.axhline(130, color='red')
# for vertical line 
plt.axvline(30, color='black')

plt.xlabel('AVG')
plt.ylabel('SR')
```

### Subplots

```python
# a diff way to plot graphs 
fig, ax = plt.subplots(figsize=(15, 7))

ax.scatter(sample_df['avg'], sample_df['strike_rate'])

ax.set_title('Something')
ax.set_xlabel('AVG')
ax.set_ylabel('SR')
```

```python
# on batters data set 
fig, ax = plt.subplots(nrows=1, ncols=2, figsize=(15,6))

ax[0].scatter(sample_df['avg'], sample_df['strike_rate'], color='red')
ax[1].scatter(sample_df['avg'], sample_df['runs'])

ax[0].set_title('AVG vs SR')
ax[0].set_xlabel('AVG')
ax[0].set_ylabel('SR')

ax[1].set_title('AVG vs Runs')
ax[1].set_xlabel('AVG')
ax[1].set_ylabel('Runs')
```

```python
fig, ax = plt.subplots(nrows=2, ncols=2, figsize=(20,12))

ax[0,0].scatter(betters['avg'], betters['strike_rate'], color='red')
ax[0,1].scatter(betters['avg'], betters['runs'])
ax[1,0].hist(betters['avg'])
ax[1,1].hist(betters['runs'], color='red')

ax[0,0].set_title('AVG vs SR')
ax[0,0].set_xlabel('AVG')
ax[0,0].set_ylabel('SR')

ax[0,1].set_title('AVG vs Runs')
ax[0,1].set_xlabel('AVG')
ax[0,1].set_ylabel('Runs')

ax[1,0].set_title('AVG Frequancy')
ax[1,0].set_xlabel('AVG')

ax[1,1].set_title('Runs Frequancy')
ax[1,1].set_xlabel('Runs')

```

### 3D Scatter Plot

```python
fig = plt.figure(figsize=(15,7))

ax = plt.subplot(projection='3d')

ax.scatter3D(betters['runs'],betters['avg'],betters['strike_rate'])

ax.set_title('IPL batsman analaysis')
ax.set_xlabel('Runs')
ax.set_ylabel('AVG')
ax.set_zlabel('SR')
```

### 3D Line Plot

```python
x = [0,1,5, 25]
y = [0,10,13, 5]
z = [0,13,20, 10]

ax = plt.subplot(projection='3d')

ax.scatter3D(x,y,z, s=[100, 100, 100, 100])
ax.plot3D(x,y,z, color='red')
```

### 3D Surface Plots

```python
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)
```

```python
xx, yy = np.meshgrid(x,y)
```

```python
z = xx**2 + yy**2
z.shape
```

```python
fig = plt.figure(figsize=(12,8))

ax = plt.subplot(projection='3d')

g = ax.plot_surface(xx,yy,z, cmap='winter')
fig.colorbar(g)
```

### Contour Plots

```python
fig = plt.figure(figsize=(12,8))

ax = plt.subplot()

g = ax.contour(xx,yy,z, cmap='winter')
fig.colorbar(g)
```

```python
fig = plt.figure(figsize=(12,8))

ax = plt.subplot()

g = ax.contourf(xx,yy,z, cmap='winter')
fig.colorbar(g)
```

### Heatmap

```python
delvires = pd.read_csv('Data Sets/IPL_Ball_by_Ball_2008_2022.csv')
delvires.head()
```

```python
sixes = delvires[(delvires['ballnumber'].isin([1,2,3,4,5,6])) & (delvires['batsman_run'] == 6)]
sixes.head()
```

```python
sixes_grid = sixes.pivot_table(
    index='overs',
    columns='ballnumber', 
    values='batsman_run', 
    aggfunc='count'
)
sixes_grid
```

```python
plt.figure(figsize=(15,10))
plt.imshow(sixes_grid, cmap='winter')

plt.yticks(delvires['overs'].unique(), list(range(1,21)))
plt.xticks(np.arange(0,6),list(range(1,7)))
plt.colorbar()
```

#### pie chart 

```python
df = pd.DataFrame(
    {
        'batsman':['Rohit', 'virat', 'sky', 'pant'], 
        'match1':[100,45,67,29],
        'match2':[120,125,69,89], 
        'match3':[10,145,47,79], 
    }
)

df
```

```python
df['match1'].plot(kind='pie', labels=df['batsman'].values, autopct='%0.2f%%', explode=[0,0,0,0.1], cmap='winter')
```

```python
# multiple pie charts
df[['match1', 'match2', 'match3']].plot(kind='pie', subplots=True, figsize=(20,6), labels=df['batsman'].values, autopct='%0.2f%%', explode=[0,0.1,0,0], cmap='winter', title='Runs Scored By Batsmans');
```

#### subplots
