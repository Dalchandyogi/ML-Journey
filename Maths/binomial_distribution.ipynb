```python
import numpy as np 
import matplotlib.pyplot as plt 
```

```python
n = 10       # number of trials
p = 0.5      # success probabilty
size = 1000  # number of samples to genrated

binomial_dist = np.random.binomial(n, p, size)
```

```python
binomial_dist
```

```python
plt.hist(binomial_dist, density=True);
```
