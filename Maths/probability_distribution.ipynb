```python
import random
import pandas as pd 
import numpy as np  
```

### Probability Distribution

#### Roll a dice 

```python
l = []
for i in range(10000):
    l.append(random.randint(1,6))
```

```python
l[:5]
```

```python
s = pd.Series(l)
s
```

```python
s = s.value_counts()/s.value_counts().sum()
s
```

```python
s.plot(kind='bar', title='PMF')
```

#### Roll 2 dice and sum of them

```python
l = []
for i in range(10000):
    d1 = random.randint(1,6)
    d2 = random.randint(1,6)
    l.append(d1 + d2)
```

```python
l[:5]
```

```python
s2 = pd.Series(l)
s2
```

```python
s2 = s2.value_counts()/s2.value_counts().sum()
s2
```

```python
s2 = s2.sort_index()
s2
```

```python
s2.plot(kind='bar', title='PMF')
```
