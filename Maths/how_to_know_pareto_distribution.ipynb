```python
import numpy as np 
import matplotlib.pyplot as plt 
```

#### 1. log(x) and log(y) 

```python
# Define parameters of pareto Distribution
alpha = 3
xm = 1
```

```python
# create an array of x value 
x = np.linspace(0.1, 10, 1000)
```

```python
# calculate y values of the pareto Distribution
y = alpha * (xm**alpha) / (x**(alpha+1))
```

```python
plt.plot(x,y)

plt.title('Pareto Distribution of x and y')
```

```python
plt.plot(np.log(x),np.log(y))

plt.title('Pareto Distribution of log(x) and log(y)')
```

#### 2. QQ Plot

```python
import scipy.stats as stats
```

```python
import statsmodels.api as sm
```

```python
# Define parameters of pareto Distribution
alpha = 2
xm = 1
```

```python
# Genrate a set of random data for pareto Distribution 
x = stats.pareto.rvs(b=alpha, scale=xm, size=1000)
```

```python
x[:7]
```

```python
plt.hist(x);
```

```python
parms = stats.pareto.fit(x, floc=0)
dist = stats.pareto(b=parms[0], scale=parms[2])
```

```python
dist
```

```python
fig = sm.qqplot(x, dist=dist, line=45)

# Add a title and labels
plt.title('QQ Plot of Pareto Distribution with Pareto fit')
plt.xlabel('Theortical Quantiles')
plt.ylabel('Sample Quantiles')

plt.show()
```
