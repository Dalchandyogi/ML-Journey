```python
# Distance from origin 

import numpy as np 

# define n dimesional vector A 
A = np.array([1, 2, 3, 4, 5])

distance = np.linalg.norm(A)

print(f'Distance from the Origin : {distance}.')
```

```python
# define two n-dimensional vectors A and B 
A = np.array([1, 2, 3, 4, 5])
B = np.array([6, 7, 8, 9, 10])

# calculate difference vectors 
difference = A - B 

difference
```

```python
# calculate the Euclidean distance between A and B 
distance = np.linalg.norm(difference)

print(f'Euclidean distance between A and B : {distance}')
```

#### Scaler opertion

```python
import matplotlib.pyplot as plt 
```

```python
# genrate random 2d data 
data = np.random.rand(100, 2)
data[:5]
```

```python
# calculate mean of data along each dimension
data_mean = np.mean(data, axis=0)

data_mean
```

```python
# performing mean centered 
centered_data = data - data_mean
centered_data
```

```python
# plot the original data
plt.figure(figsize=(15,6))

plt.subplot(1, 2, 1)
plt.scatter(data[:, 0], data[:, 1], label='Original Data')
plt.xlabel('X Axies')
plt.ylabel('Y Axies')
plt.title('Before Mean Centering')
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.legend()


# plot the mean-centered data
plt.subplot(1, 2, 2)
plt.scatter(centered_data[:, 0], centered_data[:, 1], label='Mean-Centered Data', color='orange')
plt.xlabel('X Axies')
plt.ylabel('Y Axies')
plt.title('After Mean Centering')
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.legend()
```

```python
# cosine similiarty 

# define three vectors 
A = np.arange(1, 8)
B = np.arange(-8, -1)
C = np.arange(5, 12)

print(A)
print(B)
print(C)
```

```python
# Calculate cosine similarty between A and B 
cosine_similarty_AB = np.dot(A, B) / (np.linalg.norm(A) * np.linalg.norm(A))

# Calculate cosine similarty between A and c
cosine_similarty_AC = np.dot(A, C) / (np.linalg.norm(A) * np.linalg.norm(C))
```

```python
print(f'Cosine Similarity Between A and B : {cosin_similarty_AB}.')
print(f'Cosine Similarity Between A and C : {cosin_similarty_AC}.')
```
