### Restaurant Sales Dataset(Monthly)

```python
# Columns:
# ID | January | February | March | April
sales = np.array([
    [1, 12000, 15000, 18000, 22000],
    [2, 10000, 13000, 16000, 19000],
    [3, 18000, 21000, 25000, 28000],
    [4, 9000, 11000, 14000, 17000],
    [5, 15000, 17000, 20000, 24000],
    [6, 22000, 25000, 29000, 32000],
    [7, 11000, 14000, 17000, 21000],
    [8, 16000, 19000, 23000, 27000]
])
```

```python
restaurant_ids = sales[:, 0]
restaurant_sales = sales[:, 1:]
```

```python
restaurant_sales
```

```python
# Find the total sales for each month.
# Expected result should be something like:
# January   → ?
# February  → ?
# March     → ?
# April     → ?

months = ["January", "February", "March","April"]
total_sales_each_month = np.sum(restaurant_sales, axis=0)
for i in range(0, len(total_sales_each_month)):
    print(f'{months[i]} → {total_sales_each_month[i]}')
```

```python
# Which month had the highest total sales?
np.max(total_sales_each_month)
```

```python
# Find the restaurant ID that generated the highest total sales across all 4 months.
total_sales_per_restaurant = restaurant_sales.sum(axis=1)

highest_sales_idx = np.argmax(total_sales_per_restaurant)

best_restaurant_id = restaurant_ids[highest_sales_idx]
print(best_restaurant_id)
```

```python
# Calculate the average monthly sales for each restaurant.
# Your output should contain 8 values — one for each restaurant.

np.average(restaurant_sales, axis=1)
```

```python
# Find all restaurants where April sales were greater than 25,000.
# Return the restaurant IDs.

april_sales = sales[:, 4]
mask = april_sales > 25000 
print("These Restaurant have April sales were greater than 25,000 :",restaurant_ids[mask])
```

```python
# Q6 — np.where()

# Create a new array where:

# Sales >= 20,000 → "High"
# Sales < 20,000  → "Low"

# You can apply this to the April sales column.

april_sales_category = np.where(april_sales >= 20000, "High", "Low")
print(april_sales_category)
```

```python
# Q7 — Broadcasting 🔥

# Suppose next month every restaurant gets a 10% increase in sales.

# Use broadcasting/vectorization to calculate the new April sales.

# Don't use a for loop.

new_april_sales = april_sales + (april_sales * 0.10)
print("New April Sales :", new_april_sales)
```

```python
# ⭐ Bonus

# Normalize the April sales between 0 and 1 using:

# (x - min) / (max - min)

(april_sales - np.min(april_sales)) / (np.max(april_sales) - np.min(april_sales))
```
