### Python List vs NumPy Array

```python
py_list = [1,2,3]
print("Python List Multiplication : ", py_list * 2)

np_array = np.array([1,2,3])
print("NumPY Array Multiplication : ", np_array * 2)
```

```python
import time

start = time.time()

py_list = [i*2 for i in range(10000000)]
print("List Operation time : ", time.time() - start)

start = time.time()

np_array = np.arange(10000000) * 2
print("Array Operation time : ", time.time() - start)
```

#### Mask

```python
a = np.arange(11)
```

```python
mask1 = a % 2 == 0
mask2 = a > 5
```

```python
print("Mask 1 : ", a[mask1])
print("Mask 2 : ", a[mask2])
```

### Fancy Index Vs np.where()

```python
arr = np.array([2,11,3,45,6,8,9,0])
indexes = [0,2,1] 
print(arr[indexes])
```

```python
where_res = np.where(arr > 5)
print("Where : ", where_res)
print(arr[where_res])
```

```python
condtion_arr = np.where(arr >5, "true", "false")
```

```python
print(arr)
print(condtion_arr)
```

#### combine arrays - hstack and vstack

```python
original = np.array([
    [2,3],
    [7,8]
])

new_row = [1,2]
```

```python
print("Orginal : \n", original)
print("New : \n", np.vstack((original, new_row)))  # vstack is used to add new row (vertical)
```

```python
new_col = np.array([[1],[2]])
```

```python
print("Orginal : \n", original)
print("New : \n", np.hstack((original, new_col)))  # vstack is used to add new col.
```

#### Broadcasting

```python
# Data structure: [restaurant_id, 2021, 2022, 2023, 2024] 
sales_data =np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
    [2, 120000, 140000, 160000, 190000],  # Beijing Bites
    [3, 200000, 230000, 260000, 300000],  # Pizza Hub
    [4, 180000, 210000, 240000, 270000],  # Burger Point
    [5, 160000, 185000, 205000, 230000]   # Chai Point
])
```

```python
monthly_avg = sales_data[:, 1:] / 12  # so here this 12 is devide all valuse, so this called broadcasting.
monthly_avg
```

#### work with .npy files

```python
try:
    logo = np.load('numpy logo.npy')
    
    # Display
    plt.figure(figsize=(10,5))
    plt.subplot(121)
    plt.imshow(logo)
    plt.title("NumPy Logo")

    # Dark mode
    dark_logo = 1 - logo
    plt.subplot(122)
    plt.imshow(dark_logo)
    plt.title("NumPy Logo Dark")

    
except FileNotFoundError: 
    print("File Not found!")
```
