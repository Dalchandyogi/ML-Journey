### Pandas Export 

- to csv
- to excel
- to json
- to html
- to sql

```python
import pandas as pd
```

### to_csv()

```python
df = pd.read_csv('Data Sets/deliveries.csv')
df.head()
```

```python
temp_df = df.groupby('batsman')['batsman_runs'].sum().reset_index()
temp_df
```

```python
temp_df.to_csv('Export DataSets/batsman_runs.csv', index=False)
```

### to_excel()

```python
temp_df.to_excel('Export DataSets/batsman_runs.xlsx')
```

```python
batter_vs_team = df.pivot_table(index='batsman', columns='bowling_team', values='batsman_runs', aggfunc='sum')
```

```python
with pd.ExcelWriter('Export DataSets/batsman.xlsx') as writer:
    temp_df.to_excel(writer, sheet_name='batsman_runs')
    batter_vs_team.to_excel(writer, sheet_name='batsman_vs_team')
```

### to_html()

```python
df
```

```python
sixes = df.query('batsman_runs == 6').pivot_table(index='over', columns='ball', values='batsman_runs', aggfunc='count')
sixes
```

```python
sixes.to_html('Export DataSets/sixes.html')
```

### to_json()

```python
df
```

```python
players = df.groupby(['batting_team', 'batsman'])['batsman_runs'].sum().unstack()
players
```

```python
clean_dict = players.apply(lambda col: col.dropna().to_dict()).to_dict()
clean_dict
```

```python
import json
with open('Export DataSets/players.json', 'w') as f:
    json.dump(clean_dict, f, indent=4)
```
