```python
import numpy as np 
import pandas as pd 
import re
```

```python
comapines = pd.read_csv('scrape_data/raw_companies_data.csv')
```

```python
comapines.head()
```

```python
comapines['rating_count']
```

```python
def parse_count(val):
    if pd.isna(val):
        return None

    # Remove parentheses, spaces, and commas
    clean_val = re.sub(r"[\(\)\s,]", "", str(val)).lower()

    if "k" in clean_val:
        return float(clean_val.replace("k", "")) * 1_000
    elif "l" in clean_val:
        return float(clean_val.replace("l", "")) * 100_000
    elif "m" in clean_val:
        return float(clean_val.replace("m", "")) * 1_000_000
    else:
        try:
            return float(clean_val)
        except ValueError:
            return None
```

```python
comapines['rating_count'] = comapines['rating_count'].apply(parse_count)
```

```python
comapines ['headquarter'] = comapines['location'].str.split(' ').str[0].str.strip()
```

```python
comapines['other_locations_count'] = comapines['location'].str.split(' ').str[1].str.replace('+', '').str.strip()
```

```python
comapines['reviews'] = comapines['reviews'].str.split(' ').str[0].apply(parse_count)
```

```python
comapines['salaries'] = comapines['salaries'].str.split(' ').str[0].apply(parse_count)
```

```python
comapines['interviews'] = comapines['interviews'].str.split(' ').str[0]
```

```python
comapines['jobs'] = comapines['jobs'].str.split().str[0]
```

```python
comapines['benefits'] = comapines['benefits'].str.split(' ').str[0]
```

```python
comapines['photos'] = comapines['photos'].str.split(' ').str[0]
```

```python
clean_df = comapines[['company_name', 'company_url', 'industry', 'rating', 'reviews', 'headquarter', 'other_locations_count', 'highly_rated_for', 'critically_rated_for', 'salaries', 'interviews', 'jobs', 'benefits', 'photos']]
clean_df
```

```python
clean_df.to_csv('compaines_clean_data.csv', index=False)
```
