### Configuration

````python
BASE_URL = "https://www.ambitionbox.com/list-of-companies?page={page}"

headers = {
    "User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 (KHTML, like Gecko) "
        "Chrome/80.0.3987.162 Safari/537.36"
    )
}

MAX_WORKERS = 5
CHECKPOINT_EVERY = 50

# Use a separate directory so this final checkpoint system
# does not conflict with the older CSV checkpoint.
CHECKPOINT_DIR = "scrape_data"
CHECKPOINT_FILE = os.path.join(
    CHECKPOINT_DIR, "raw_companies_data.csv"
)
STATE_FILE = os.path.join(
    CHECKPOINT_DIR, "scrape_state.json"
)
````

### HTTP Session

````python
# A separate Session is created for each worker thread.
thread_local = threading.local()


def get_session():
    if not hasattr(thread_local, "session"):
        session = requests.Session()
        session.headers.update(headers)
        thread_local.session = session

    return thread_local.session
````

### Fetch Page with Retry

````python
def fetch_page(page, retries=3, timeout=15):
    url = BASE_URL.format(page=page)

    for attempt in range(1, retries + 1):
        try:
            response = get_session().get(
                url=url,
                timeout=timeout
            )

            # Retry temporary rate-limit/server errors.
            if response.status_code == 429 or 500 <= response.status_code < 600:
                if attempt == retries:
                    response.raise_for_status()

                retry_after = response.headers.get("Retry-After")

                try:
                    wait = float(retry_after)
                except (TypeError, ValueError):
                    wait = min(2 ** (attempt - 1), 8)

                time.sleep(wait)
                continue

            response.raise_for_status()
            return response

        except requests.RequestException:
            if attempt == retries:
                raise

            # Exponential backoff: 1s, 2s, ...
            time.sleep(min(2 ** (attempt - 1), 8))

    raise RuntimeError(
        f"Page {page} failed after {retries} attempts"
    )
````

### Parse Page

````python
def parse_page(response):
    return BeautifulSoup(response.text, "lxml")


def clean_text(element):
    return element.get_text(" ", strip=True) if element else None
````

### Parse One Company Card

````python
def parse_company(card):
    company = {}

    # Company name
    name = card.find("h2")
    company["company_name"] = clean_text(name)

    # Company overview URL available in the listing card metadata
    url_meta = card.find("meta", itemprop="url")
    company["company_url"] = (url_meta.get("content") if url_meta else None)

    # Rating
    rating = card.find("div", class_="rating_text")
    company["rating"] = clean_text(rating)

    # Rating count, e.g. "(1.2L)"
    rating_count = card.find("span",class_="companyCardWrapper__companyRatingCount")
    company["rating_count"] = clean_text(rating_count)

    # Industry + location
    interlinking = card.find("span",class_="companyCardWrapper__interLinking")

    if interlinking:
        parts = [
            part.strip()
            for part in interlinking.get_text("|", strip=True).split("|")
        ]

        company["industry"] = parts[0] if parts else None
        company["location"] = parts[1] if len(parts) > 1 else None
    else:
        company["industry"] = None
        company["location"] = None

    # Reviews, salaries, interviews, jobs, benefits, photos
    # mapped by visible title rather than position.
    actions = {}

    for action in card.find_all("a", class_="companyCardWrapper__ActionWrapper"):
        title = action.find("span",class_="companyCardWrapper__ActionTitle")

        if title:
            key = clean_text(title).lower()
            actions[key] = clean_text(action)

    for key in [
        "reviews",
        "salaries",
        "interviews",
        "jobs",
        "benefits",
        "photos"
    ]:
        company[key] = actions.get(key)

    # Highly Rated For / Critically Rated For
    company["highly_rated_for"] = None
    company["critically_rated_for"] = None

    comparison_blocks = card.find_all("div",class_="companyCardWrapper__ratingComparisonWrapper")

    for block in comparison_blocks:
        header = block.find(
            "span",
            class_=lambda value: (
                value
                and "companyCardWrapper__ratingHeader" in value
            )
        )

        values = block.find(
            "span",
            class_="companyCardWrapper__ratingValues"
        )

        header_text = clean_text(header)
        values_text = clean_text(values)

        if not header_text or not values_text:
            continue

        if "Highly Rated For" in header_text:
            company["highly_rated_for"] = values_text

        elif "Critically Rated For" in header_text:
            company["critically_rated_for"] = values_text

    return company
````

### Scrape One Page

````python
def scrape_page(page):
    response = fetch_page(page)
    soup = parse_page(response)

    company_cards = soup.find_all(
        "div",
        class_="companyCardWrapper"
    )

    scraped_at = datetime.now(timezone.utc).date().isoformat()

    companies = []

    for card in company_cards:
        company = parse_company(card)

        company["page_number"] = page
        company["scraped_at"] = scraped_at

        companies.append(company)

    return companies
````

### Checkpoint Functions

````python
def save_checkpoint(
    companies,
    completed_pages,
    failed_pages
):
    os.makedirs(CHECKPOINT_DIR, exist_ok=True)

    pd.DataFrame(companies).to_csv(
        CHECKPOINT_FILE,
        index=False
    )

    state = {
        "completed_pages": sorted(completed_pages),
        "failed_pages": sorted(failed_pages),
        "saved_at": datetime.now(timezone.utc).isoformat()
    }

    with open(STATE_FILE, "w", encoding="utf-8") as file:
        json.dump(state, file, indent=2)


def load_checkpoint():
    companies = []
    completed_pages = set()
    failed_pages = set()

    if os.path.exists(CHECKPOINT_FILE):
        companies = pd.read_csv(
            CHECKPOINT_FILE
        ).to_dict("records")

    if os.path.exists(STATE_FILE):
        with open(STATE_FILE, "r", encoding="utf-8") as file:
            state = json.load(file)

        completed_pages = set(
            state.get("completed_pages", [])
        )

        failed_pages = set(
            state.get("failed_pages", [])
        )

    return companies, completed_pages, failed_pages
````

### Concurrent Scraping with Live Progress

````python
def scrape_pages(
    pages,
    all_companies,
    completed_pages,
    failed_pages,
    max_workers=5,
    checkpoint_every=50
):
    pages = list(pages)

    if not pages:
        return (
            all_companies,
            completed_pages,
            failed_pages
        )

    completed_since_checkpoint = 0

    with ThreadPoolExecutor(
        max_workers=max_workers
    ) as executor:

        future_to_page = {
            executor.submit(scrape_page, page): page
            for page in pages
        }

        with tqdm(
            total=len(pages),
            desc="Scraping pages",
            unit="page"
        ) as progress:

            for future in as_completed(future_to_page):
                page = future_to_page[future]

                try:
                    companies = future.result()

                    all_companies.extend(companies)
                    completed_pages.add(page)
                    failed_pages.discard(page)

                except Exception:
                    failed_pages.add(page)

                progress.update(1)

                progress.set_postfix(
                    companies=len(all_companies),
                    failed=len(failed_pages)
                )

                completed_since_checkpoint += 1

                if completed_since_checkpoint >= checkpoint_every:
                    save_checkpoint(
                        all_companies,
                        completed_pages,
                        failed_pages
                    )
                    completed_since_checkpoint = 0

    save_checkpoint(
        all_companies,
        completed_pages,
        failed_pages
    )

    return (
        all_companies,
        completed_pages,
        failed_pages
    )
````

### Retry Failed Pages

Pages that still fail after the request-level retries are retried in separate rounds. The retry progress is shown with a progress bar instead of printing one line for every page.


````python
def retry_failed_pages(
    all_companies,
    completed_pages,
    failed_pages,
    max_workers=5,
    checkpoint_every=50,
    max_rounds=2
):
    for round_number in range(1, max_rounds + 1):

        pages_to_retry = sorted(failed_pages)

        if not pages_to_retry:
            break

        print(
            f"Retry round {round_number}: "
            f"{len(pages_to_retry)} failed pages"
        )

        before_completed = set(completed_pages)

        (
            all_companies,
            completed_pages,
            failed_pages
        ) = scrape_pages(
            pages_to_retry,
            all_companies,
            completed_pages,
            failed_pages,
            max_workers=max_workers,
            checkpoint_every=checkpoint_every
        )

        newly_completed = (
            completed_pages - before_completed
        )

        if not newly_completed:
            break

    return (
        all_companies,
        completed_pages,
        failed_pages
    )
````

### Final Validation

````python
def validate_data(
    companies,
    start_page,
    end_page,
    completed_pages,
    failed_pages
):
    df = pd.DataFrame(companies)

    expected_pages = set(
        range(start_page, end_page + 1)
    )

    missing_pages = (
        expected_pages - completed_pages
    )

    print("\nFinal validation")
    print(
        f"Pages completed : "
        f"{len(completed_pages)}/{len(expected_pages)}"
    )
    print(
        f"Failed pages    : "
        f"{len(failed_pages)}"
    )
    print(
        f"Companies       : "
        f"{len(df)}"
    )

    if "company_name" in df.columns:
        print(
            f"Missing names   : "
            f"{df['company_name'].isna().sum()}"
        )
        print(
            f"Duplicate names : "
            f"{df['company_name'].duplicated().sum()}"
        )
    else:
        print("Missing names   : N/A")
        print("Duplicate names : N/A")

    if missing_pages:
        print(
            f"Missing pages   : "
            f"{sorted(missing_pages)}"
        )
    else:
        print("Missing pages   : 0")

    return df
````

### Main Function

`main()` controls the complete workflow:

1. load checkpoint when `resume=True`
2. determine pages that still need scraping
3. scrape concurrently
4. save checkpoints
5. retry failed pages
6. validate the result
7. write `raw_company_data.csv`


````python
def main(
    start_page=1,
    end_page=500,
    max_workers=5,
    checkpoint_every=50,
    resume=True,
    retry_rounds=2
):
    start_time = time.perf_counter()

    if resume:
        (all_companies,completed_pages,failed_pages) = load_checkpoint()
    else:
        all_companies = []
        completed_pages = set()
        failed_pages = set()

    target_pages = set(range(start_page, end_page + 1))

    completed_pages &= target_pages
    failed_pages &= target_pages

    pages_to_scrape = sorted(
        target_pages - completed_pages
    )

    print(
        f"Starting scrape | "
        f"pages={start_page}-{end_page} | "
        f"remaining={len(pages_to_scrape)} | "
        f"workers={max_workers}"
    )

    if pages_to_scrape:
        (all_companies,completed_pages,failed_pages) = scrape_pages(
            pages_to_scrape,
            all_companies,
            completed_pages,
            failed_pages,
            max_workers=max_workers,
            checkpoint_every=checkpoint_every
        )

    if failed_pages:
        (
            all_companies,
            completed_pages,
            failed_pages
        ) = retry_failed_pages(
            all_companies,
            completed_pages,
            failed_pages,
            max_workers=max_workers,
            checkpoint_every=checkpoint_every,
            max_rounds=retry_rounds
        )

    df = validate_data(
        all_companies,
        start_page,
        end_page,
        completed_pages,
        failed_pages
    )

    df.to_csv(
        "raw_company_data.csv",
        index=False
    )

    save_checkpoint(
        all_companies,
        completed_pages,
        failed_pages
    )

    elapsed = time.perf_counter() - start_time

    print(f"\nSaved: raw_company_data.csv")
    print(f"Time elapsed: {elapsed:.1f} seconds")

    if failed_pages:
        print(
            f"Still failed: {sorted(failed_pages)}"
        )
    else:
        print("Scraping completed successfully.")

    return df, sorted(failed_pages)
````

### Resume an Interrupted 500-Page Scrape

If the 500-page run is interrupted after the new checkpoint system has saved state, run:

```python
df, failed_pages = main(
    start_page=1,
    end_page=500,
    max_workers=5,
    checkpoint_every=50,
    resume=True,
    retry_rounds=2
)
```

Completed pages recorded in `scrape_checkpoint/scrape_state.json` will be skipped.
