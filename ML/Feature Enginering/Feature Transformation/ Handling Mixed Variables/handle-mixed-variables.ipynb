### Handling Mixed Variables

A **mixed variable** is a feature that contains both numerical and categorical information in the same value. Before using it in an ML model, split the value into meaningful numerical and categorical features.

**Why do this ?**
- Numerical parts can be used as quantities or measurements.
- Categorical parts can represent groups, types, or labels.
- Separating them allows us to apply suitable preprocessing to each part.

In this notebook, the Titanic dataset provides three examples: `number`, `Cabin`, and `Ticket`.

```python
import numpy as np 
import pandas as pd 
```

```python
df = pd.read_csv("Datasets/titanic_toy.csv")

df.head()
```

#### 1. Separate numerical and categorical parts

The `number` column contains values such as `5`, `3`, and `A`. Convert values that can be interpreted as numbers using `pd.to_numeric(errors='coerce')`. Values that cannot be converted become `NaN`.

```python
df['number'].unique()
```

```python
fig = df['number'].value_counts().plot.bar()
fig.set_title('Passengers travelling with')
```

```python
# extract numerical part
df['number_numerical'] = pd.to_numeric(df["number"],errors='coerce',downcast='integer')
```

```python
# extract categorical part
df['number_categorical'] = np.where(df['number_numerical'].isnull(),df['number'],np.nan)

df.head()
```

#### 2. Extract information from Cabin

A cabin value such as `C85` contains:
- `C`: the cabin's starting letter (categorical part).
- `85`: the numerical part.

```python
df['Cabin'].unique()
```

```python
df['Ticket'].unique()
```

```python
df['cabin_num'] = df['Cabin'].str.extract('(\d+)') # captures numerical part
df['cabin_cat'] = df['Cabin'].str[0] # captures the first letter

df.head()
```

#### 3. Extract information from Ticket

Ticket values can contain a prefix and a number, for example `A/5 21171`.

- `split()[-1]` selects the last part, which is converted to a number in `ticket_num`.
- `split()[0]` selects the first part as a possible category in `ticket_cat`.

```python
# extract the last bit of ticket as number
df['ticket_num'] = df['Ticket'].apply(lambda s: s.split()[-1])
df['ticket_num'] = pd.to_numeric(df['ticket_num'],
                                   errors='coerce',
                                   downcast='integer')

# extract the first part of ticket as category
df['ticket_cat'] = df['Ticket'].apply(lambda s: s.split()[0])
df['ticket_cat'] = np.where(df['ticket_cat'].str.isdigit(), np.nan,
                              df['ticket_cat'])

df.head(20)
```

```python
df['ticket_cat'].unique()
```

### Key takeaways

- Inspect unique values before deciding how to split a mixed feature.
- Extract numeric and categorical information into separate columns.
- Check the extracted columns for missing values and unexpected formats.
- Fit any later preprocessing steps using training data only, then apply them to validation/test data.

**Goal:** turn one mixed column into separate features that can be handled with appropriate numerical and categorical preprocessing.
