### Handling Date and Time Features

Date and time columns often contain useful patterns, such as seasonality, day of the week, time of day, or how recently an event happened. ML models generally work more effectively when these values are converted into meaningful numerical features.

In Pandas, `pd.to_datetime()` converts a column to a datetime type, and the `.dt` accessor extracts date/time components.

```python
import numpy as np
import pandas as pd
```

```python
orders = pd.read_csv('Datasets/orders.csv')
messages = pd.read_csv('Datasets/messages.csv')
```

### Working with Dates

**Convert a column to datetime**

`pd.to_datetime()` converts date-like values into Pandas datetime values. This is important because string columns do not support datetime operations such as `.dt.year` or `.dt.month`.

After conversion, use the `.dt` accessor to extract date components.

```python
# Converting to datetime datatype
orders['date'] = pd.to_datetime(orders['date'])
```

#### Extract calendar features

The following cells extract parts of a date:

- **Year :** `.dt.year`
- **Month :** `.dt.month` gives the month number; `.dt.month_name()` gives its name.
- **Day :** `.dt.day` gives the day of the month.
- **Day of week :** `.dt.dayofweek` uses Monday = 0 and Sunday = 6; `.dt.day_name()` returns the weekday name.
- **Weekend indicator :** `np.where()` creates a binary feature (1 for Saturday/Sunday, otherwise 0).
- **Week number :** `.dt.isocalendar().week` returns the ISO week number.
- **Quarter :** `.dt.quarter` returns 1 to 4.
- **Semester :** the notebook groups quarters 1–2 as semester 1 and quarters 3–4 as semester 2.

These features can help a model learn calendar-related patterns. Keep only features that make sense for the problem.

```python
orders['date_year'] = orders['date'].dt.year

orders.sample(5)
```

```python
orders['date_month_no'] = orders['date'].dt.month

orders.head()
```

```python
orders['date_month_name'] = orders['date'].dt.month_name()

orders.head()
```

```python
orders['date_day'] = orders['date'].dt.day

orders.head()
```

```python
# day of week
orders['date_dow'] = orders['date'].dt.dayofweek

orders.head()
```

```python
# day of week - name

orders['date_dow_name'] = orders['date'].dt.day_name()

orders.head()
```

```python
# is weekend?

orders['date_is_weekend'] = np.where(orders['date_dow_name'].isin(['Sunday', 'Saturday']), 1,0)

orders.head()
```

```python
orders['date_week'] = orders['date'].dt.isocalendar().week
orders.head()
```

```python
orders['quarter'] = orders['date'].dt.quarter
```

```python
orders.head()
```

```python
orders['semester'] = np.where(orders['quarter'].isin([1,2]), 1, 2)
```

```python
orders.head()
```

#### Extract Time elapsed between dates

```python
import datetime

today = datetime.datetime.today()

today
```

```python
today - orders['date']
```

```python
(today - orders['date']).dt.days
```

```python
# Elapsed days (rounded)
np.round((today - orders['date']) / np.timedelta64(1, 'D'),0)
```

### Working with Time

```python
messages.info()
```

```python
# Converting to datetime datatype
messages['date'] = pd.to_datetime(messages['date'])
```

```python
messages.info()
```

#### Extract hour, minute, and second

- `.dt.hour` extracts the hour (0–23).
- `.dt.minute` extracts the minute (0–59).
- `.dt.second` extracts the second (0–59).

These features can reveal time-of-day patterns, such as message or transaction activity.

```python
# Extract hour, time, second 
messages['hour'] = messages['date'].dt.hour
messages['min'] = messages['date'].dt.minute
messages['sec'] = messages['date'].dt.second

messages.head()
```

#### Extract Time part

```python
messages['time'] = messages['date'].dt.time
messages.head()
```

#### Time difference

```python
today - messages['date']
```

```python
# in seconds
(today - messages['date'])/np.timedelta64(1,'s')
```

```python
# in minutes
(today - messages['date'])/np.timedelta64(1,'m')
```

```python
# elapsed time in hours
(today - messages['date'])/np.timedelta64(1,'h')
```

### Key Takeaways

- Convert date strings with `pd.to_datetime()` before extracting components.
- Use `.dt` for date/time feature extraction.
- Derive calendar features (year, month, weekday, weekend, quarter) when they may explain the target.
- Use `Timedelta` arithmetic to calculate elapsed durations.
- Check missing or invalid dates and avoid using future information when creating time-based features.
