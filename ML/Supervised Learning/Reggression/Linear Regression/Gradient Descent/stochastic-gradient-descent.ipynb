### Stochastic Gradient Descent (SGD)

**Stochastic Gradient Descent (SGD)** is an optimization algorithm that trains a machine learning model by updating its parameters using **one training example at a time**. It aims to minimize the model's loss (cost) function.

#### Why do we use SGD?

- **Frequent updates:** Parameters are updated after each training example.
- **Memory efficient:** It does not need to calculate each update using the entire dataset.
- **Useful for large datasets:** The model can begin learning without processing all rows for every update.
- **Noisy updates:** Individual updates fluctuate, which can sometimes help the optimizer move through difficult regions. However, training may be less stable than Batch Gradient Descent.

#### How does SGD work?

1. Initialize the model's coefficients and intercept.
2. Select one training example.
3. Calculate the prediction for that example.
4. Calculate the loss and its gradient for that example.
5. Update the parameters using the learning rate.
6. Repeat for the training examples over multiple epochs.

For a parameter $\theta$, the update rule is:

$$
\theta_{new} = \theta_{old} - \eta \nabla J_i(\theta)
$$

Where:

- $\theta_{\text{new}}$ = Updated parameter
- $\theta_{\text{old}}$ = Previous parameter
- $\eta$ = Learning rate
- $\nabla J_i(\theta)$ = Gradient calculated using one training example

### Prepare the data and establish a baseline model

```python
from sklearn.datasets import load_diabetes

import numpy as np 

from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split
```

```python
X, y = load_diabetes(return_X_y=True)
```

```python
print(X.shape)
print(y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.2, random_state=2)
```

```python
reg = LinearRegression()
```

```python
reg.fit(X_train, y_train)
```

```python
print('coef_ :', reg.coef_, '\n')
print('intercept_ :', reg.intercept_)
```

```python
y_pred = reg.predict(X_test)
print('r2 score : ', r2_score(y_test, y_pred))
```

### Implement SGD from Scratch

```python
class SGDRegressor: 
    
    def __init__(self, learning_rate=0.01, epochs=100): 
        self.coef_ = None
        self.intercept_ = None
        self.lr = learning_rate
        self.epochs = epochs

    def fit(self, X_train, y_train): 

        # set random coefs and intercept values
        self.intercept_ = 0 
        self.coef_ = np.ones(X_train.shape[1])

        for i in range(self.epochs): 
            for j in range(X_train.shape[0]):
                idx = np.random.randint(0, X_train.shape[0])

                y_hat = np.dot(X_train[idx], self.coef_) + self.intercept_
                
                slope_intercept = -2 * (y_train[idx] - y_hat)
                self.intercept_ = self.intercept_ - (self.lr * slope_intercept)

                slope_coef = -2 * np.dot((y_train[idx] - y_hat), X_train[idx])
                self.coef_ = self.coef_ - (self.lr * slope_coef)
                

        print('intercept_ :', self.intercept_)
        print('coefs_ :', self.coef_)

    def predict(self, X_test): 
        return np.dot(X_test, self.coef_) + self.intercept_
    
```

```python
sgd = SGDRegressor(learning_rate=0.01, epochs=50)
```

```python
sgd.fit(X_train, y_train)
```

```python
y_pred = sgd.predict(X_test)
```

```python
r2_score(y_test, y_pred)
```

### Observe variation across repeated runs - `Its give different results at different time`

```python
sgd1 = SGDRegressor(learning_rate=0.01, epochs=50)
sgd1.fit(X_train, y_train)
y_pred = sgd1.predict(X_test)
print('r2_score :', r2_score(y_test, y_pred))
```

```python
sgd2 = SGDRegressor(learning_rate=0.01, epochs=50)
sgd2.fit(X_train, y_train)
y_pred = sgd2.predict(X_test)
print('\nr2_score :', r2_score(y_test, y_pred))
```

```python
sgd3 = SGDRegressor(learning_rate=0.01, epochs=50)
sgd3.fit(X_train, y_train)
y_pred = sgd3.predict(X_test)
print('\nr2_score :', r2_score(y_test, y_pred))
```

```python
sgd4 = SGDRegressor(learning_rate=0.01, epochs=50)
sgd4.fit(X_train, y_train)
y_pred = sgd4.predict(X_test)
print('\nr2_score :', r2_score(y_test, y_pred))
```

### Key Takeaways

1. SGD updates model parameters using one training example at a time.
2. The learning rate controls the size of each parameter update.
3. An epoch is one pass through the training data; this implementation makes repeated single-example updates during each epoch.
4. SGD updates are noisy, so results can vary between runs.
5. Feature scaling, learning rate, number of epochs, and random initialization can affect training.
6. Scikit-learn's `SGDRegressor` provides a ready-to-use implementation with additional configuration options.

**Remember:** SGD is an optimization algorithm, not a regression model. It can be used to train suitable models by minimizing their loss functions.
